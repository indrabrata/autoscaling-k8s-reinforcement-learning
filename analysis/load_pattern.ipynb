# K6 Training Load Pattern (Single Cycle - 30 Minutes)

Visualization of the ramping VUs pattern used in `k6-train-patterns-loop.js`.

```python
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.dpi'] = 100

stages = [
    {"duration_min": 6, "target": 10},
    {"duration_min": 6, "target": 40},
    {"duration_min": 6, "target": 75},
    {"duration_min": 6, "target": 30},
    {"duration_min": 6, "target": 5},
]

times = [0]
vus = [0]
t = 0
for stage in stages:
    t += stage["duration_min"]
    times.append(t)
    vus.append(stage["target"])

fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(times, vus, marker='o', linewidth=2.5, color='#2196F3', markersize=8, zorder=3)
ax.fill_between(times, vus, alpha=0.15, color='#2196F3')

for x, y in zip(times, vus):
    ax.annotate(f'{y} VUs', xy=(x, y), xytext=(0, 10),
                textcoords='offset points', ha='center', fontsize=10, fontweight='bold')

ax.set_xlabel('Time (minutes)')
ax.set_ylabel('Virtual Users (VUs)')
ax.set_title('K6 Training Load Pattern — Single Cycle (30 min)')
ax.set_xticks(times)
ax.set_xticklabels([])
ax.set_ylim(-15, 120)
ax.set_xlim(-1, 31)

plt.tight_layout()
plt.show()
```
